Get all relevant EKP nan samples

```python
import os
import pandas as pd

csv_datei = "/groups/ds/Win-KID/BVBRC/BVBRC_genome_amr_02.2025.csv"
df = pd.read_csv(csv_datei, sep=",")

nan_rows = df[df["Laboratory Typing Platform"].isna()]
print(
    f"Unique genome IDs with NaN in 'Laboratory Typing Platform': {len(nan_rows["Genome ID"].unique())}"
)
non_nan_rows = df[~df["Laboratory Typing Platform"].isna()]

# Use only samples that are not in devices dataset
nan_rows = nan_rows[
    ~nan_rows["Genome ID"].isin(non_nan_rows["Genome ID"])
]

# Filter out EKP
nan_rows["Species"] = (
    nan_rows["Genome Name"].astype(str).str.split().str[:2].str.join(" ")
)
nan_EKP_df = nan_rows[
    nan_rows["Species"].str.contains("Klebsiella pneumoniae", case=False, na=False)
]

# Filter out samples with fna
fna_folder = "/groups/ds/Win-KID/BVBRC/fna"
existing_ids = [
        gid
        for gid in nan_EKP_df["Genome ID"].unique()
        if os.path.exists(os.path.join(fna_folder, f"{gid}.fna.gz"))
    ]

print("Number of existing_ids nan EKP Genomes:")
print(len(existing_ids))
existing_ids_df = pd.DataFrame(existing_ids, columns=["Genome ID"])
existing_ids_df.to_csv("exisisting_nan_ids.csv", index=False)

nan_EKP_df = nan_EKP_df[
    nan_EKP_df["Genome ID"].isin(existing_ids_df["Genome ID"])
]

nan_EKP_df.to_csv("/groups/ds/Win-KID/BVBRC/BVBRC_devices/NA/EKP/existing_na.csv", index=False)
```

Create and save splits

```python
import pandas as pd
from pathlib import Path

N_REPEATS = 5

OUTPUT_DIR = Path("/groups/ds/Win-KID/BVBRC/BVBRC_devices/NA/EKP/CV_Conclusio")

na_df = pd.read_csv("/groups/ds/Win-KID/BVBRC/BVBRC_devices/NA/EKP/CV/na_cleaned.csv")

vitek_df = pd.read_csv(
    "/groups/ds/Win-KID/BVBRC/BVBRC_devices/NA/EKP/CV/VITEK_cleaned.csv"
)

phoenix_df = pd.read_csv(
    "/groups/ds/Win-KID/BVBRC/BVBRC_devices/NA/EKP/CV/phoenix_cleaned.csv"
)


for repeat in range(N_REPEATS):

    seed = 42 + repeat

    run_dir = OUTPUT_DIR / f"repeat{repeat}"
    run_dir.mkdir(parents=True, exist_ok=True)

    # ==================================================
    # PHOENIX
    # ==================================================

    phoenix_test = phoenix_df.sample(n=300, random_state=seed)

    phoenix_remaining = phoenix_df.drop(phoenix_test.index)

    phoenix_train_1200 = phoenix_remaining.sample(n=1200, random_state=seed)

    phoenix_train_400 = phoenix_train_1200.sample(n=400, random_state=seed + 1000)

    # ==================================================
    # VITEK
    # ==================================================

    vitek_test = vitek_df.sample(n=300, random_state=seed)

    vitek_remaining = vitek_df.drop(vitek_test.index)

    vitek_train_400 = vitek_remaining.sample(n=400, random_state=seed)

    # ==================================================
    # NA
    # ==================================================

    na_test = na_df.sample(n=300, random_state=seed)

    na_remaining = na_df.drop(na_test.index)

    na_train_400 = na_remaining.sample(n=400, random_state=seed)

    # ==================================================
    # SAVE
    # ==================================================

    phoenix_train_1200.to_csv(
        run_dir / "phoenix_train_1200.csv",
        index=False,
    )

    phoenix_train_400.to_csv(
        run_dir / "phoenix_train_400.csv",
        index=False,
    )

    vitek_train_400.to_csv(
        run_dir / "vitek_train_400.csv",
        index=False,
    )

    na_train_400.to_csv(
        run_dir / "na_train_400.csv",
        index=False,
    )

    phoenix_test.to_csv(
        run_dir / "phoenix_test_300.csv",
        index=False,
    )

    vitek_test.to_csv(
        run_dir / "vitek_test_300.csv",
        index=False,
    )

    na_test.to_csv(
        run_dir / "na_test_300.csv",
        index=False,
    )

    print(f"Saved: {run_dir}")
```
