Get metadata for Experiment 1 samples

```python
import pandas as pd

common_samples = pd.read_csv(
    "/groups/ds/Win-KID/BVBRC/BVBRC_devices/git/stackPredAMR_Training_Data_Composition/Experiment_1/input_data/vitek_filtered.csv",
    sep=",",
)
full_df = pd.read_csv("/groups/ds/Win-KID/BVBRC/BVBRC_genome_amr_02.2025.csv", sep=",")

common_samples_meta = full_df[
    full_df["Genome ID"].astype(str).isin(common_samples["Sample_ID_IfH"].astype(str))
]

common_samples_meta.to_csv("meta_output.csv", index=False)
```

Get assosiated PubMed entries

```python
common_samples = pd.read_csv(
    "meta_output.csv",
    sep=",",
)
pubmed_entry_counts = (
    common_samples.groupby("PubMed")["Genome ID"].nunique().sort_values(ascending=False)
)
print(pubmed_entry_counts)
```

Match Genome ID with Genome Name

```python
common_samples = pd.read_csv(
    "meta_output.csv",
    sep=",",
)

df_reduced = (
    common_samples_meta[
        ["Genome ID", "Genome Name", "Laboratory Typing Platform", "PubMed"]
    ]
    .dropna(subset=["Laboratory Typing Platform"])
    .drop_duplicates()
    .sort_values("Genome ID")
    .reset_index(drop=True)
)
df_reduced.to_csv("reduced_output.csv", index=False)

genome_name_counts = df_reduced.groupby("Genome ID")["Genome Name"].nunique(
    dropna=False
)

matching_genomes = (genome_name_counts == 1).sum()
non_matching_genomes = (genome_name_counts > 1).sum()

print(f"Matching genomes: {matching_genomes}")
print(f"Non-matching genomes: {non_matching_genomes}")
```
