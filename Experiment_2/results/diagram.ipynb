```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

mean_df = pd.read_csv("diagram_data/phoenix_prediction_results.csv", index_col=0)
std_df = pd.read_csv("diagram_data/phoenix_prediction_std.csv", index_col=0)

x = np.arange(len(mean_df.index))
width = 0.25

fig, ax = plt.subplots(figsize=(6.5, 4.5))

bars1 = ax.bar(
    x - width,
    mean_df["Acc"],
    width,
    yerr=std_df["Acc"],
    capsize=4,
    label="Accuracy",
    color="#D81B60",
)

bars2 = ax.bar(
    x,
    mean_df["ROC AUC"],
    width,
    yerr=std_df["ROC AUC"],
    capsize=4,
    label="ROC AUC",
    color="#1E88E5",
)

bars3 = ax.bar(
    x + width,
    mean_df["F1-Score"],
    width,
    yerr=std_df["F1-Score"],
    capsize=4,
    label="F1-Score",
    color="#FFC107",
)

for bars in [bars1, bars2, bars3]:
    ax.bar_label(
        bars,
        fmt="%.2f",
        padding=3,
        fontsize=9,
    )

ax.set_xticks(x)
ax.set_xticklabels(mean_df.index, fontsize=11)
ax.set_ylabel("Performance", fontsize=12)
ax.set_ylim(0.5, 1.0)

ax.grid(axis="y", linestyle="--", alpha=0.4)
ax.set_axisbelow(True)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.legend(
    loc="upper center",
    bbox_to_anchor=(0.5, -0.15),
    ncol=3,
    frameon=False,
)
plt.tight_layout()

plt.savefig(
    "diagram_data/performance_comparison_phoenix.pdf", dpi=600, bbox_inches="tight"
)
plt.savefig(
    "diagram_data/performance_comparison_phoenix.jpg", dpi=600, bbox_inches="tight"
)

plt.show()
```
