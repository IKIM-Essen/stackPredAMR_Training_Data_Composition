```python
import pandas as pd
from pathlib import Path
import numpy as np


def save_mean(file_name):

    print("Experiment_4/stackPredAMR_Results/Evaluation_" + file_name + "_fold*/Evaluation.csv")
    # Collect all evaluation files
    base_dir = Path("")
    # file_name = "trainNa"
    files = list(base_dir.glob("Evaluation_" + file_name + "_fold*/Evaluation.csv"))

    dfs = [pd.read_csv(f) for f in files]

    # -----------------------------
    # Check: AB column consistency
    # -----------------------------
    ab_ref = dfs[0].iloc[:, 0]

    for i, df in enumerate(dfs[1:], start=1):
        if not ab_ref.equals(df.iloc[:, 0]):
            raise ValueError(f"AB column mismatch in file index {i}")

    # -----------------------------
    # Extract metric columns
    # -----------------------------
    metric_cols = dfs[0].columns[1:]

    # Stack values: shape = (n_runs, n_rows, n_metrics)
    stacked = np.stack([df.iloc[:, 1:].values for df in dfs], axis=0)

    # -----------------------------
    # Compute mean and std
    # -----------------------------
    mean_vals = stacked.mean(axis=0)
    std_vals = stacked.std(axis=0)

    # -----------------------------
    # Build result DataFrames
    # -----------------------------
    mean_df = pd.DataFrame(mean_vals, columns=metric_cols)
    std_df = pd.DataFrame(std_vals, columns=metric_cols)

    # Add AB column
    mean_df.insert(0, dfs[0].columns[0], ab_ref.reset_index(drop=True))
    std_df.insert(0, dfs[0].columns[0], ab_ref.reset_index(drop=True))

    # -----------------------------
    # Save results
    # -----------------------------
    mean_df.to_csv("Evaluation_mean_" + file_name + ".csv", index=False)
    std_df.to_csv("Evaluation_std_" + file_name + ".csv", index=False)


save_mean("trainNa_test")
save_mean("trainPhoenix_test")
save_mean("trainVitek_test")
save_mean("trainNa")
save_mean("trainPhoenix")
```
